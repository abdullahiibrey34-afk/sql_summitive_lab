### Step 4: USA Credit and Inventory Policy - Visual Required
The USA based product team is planning to adjust its credit policies and inventory allocation strategy based on the average credit limit of its customers. They would like to target this strategy at a state level with several goals in mind.

    - States with higher average credit limits might be able to place larger orders, justifying priority in inventory allocation.

```python
# Replace None with your code
df_state_credit = pd.read_sql("""
SELECT state, AVG(CAST(creditLimit AS REAL)) AS avg_credit_limit
FROM customers
WHERE country = 'USA'
GROUP BY state
ORDER BY avg_credit_limit DESC
""", conn)
df_state_credit
```

### Step 6: Top Customer + Product Quantities
The product team is running an analysis on popular and common products sold to each customer in order to try and determine what new products they should be looking at to include in their catalog. This data will also be used by individual sales reps to recommend similar products to each customer next time they place an order. 

They have asked you to query information, for each customer, about any product they have purchased 10 or more units of. In addition they would like the full set of data to be sorted in ascending order by the total amount purchased.

Execute a SQl query that returns the information required to address this ask.

Hint: For this one, you'll need to make use of HAVING, GROUP BY, and ORDER BY — make sure you get the order of them correct!

```python
# Replace None with you code
df_customer_products = pd.read_sql("""
SELECT c.customerName, p.productName, p.productCode,
       SUM(od.quantityOrdered) AS total_units,
       SUM(od.quantityOrdered * od.priceEach) AS total_amount
FROM customers c
JOIN orders o USING(customerNumber)
JOIN orderdetails od USING(orderNumber)
JOIN products p USING(productCode)
GROUP BY c.customerNumber, c.customerName, p.productCode, p.productName
HAVING SUM(od.quantityOrdered) >= 10
ORDER BY total_amount ASC
""", conn)
df_customer_products
```

### Step 7: Product Analysis - Visual Required

The product team is looking into the demand across its different product lines. They are conducting a comprehensive review of its product portfolio and inventory management strategies. You have been asked to query data pertaining to each different product line, that contains the total quantity ordered and the total number of products for each respective product line. By examining the number of products and total quantity ordered for each product line, the company aims to:
1. Optimize product mix:
    - Identify which product lines have the most diverse offerings (high number of products)
    - Determine which lines are most popular (high total quantity ordered)
    - Compare if lines with more products necessarily lead to more orders
2. Improve inventory management:
    - Adjust stock levels based on the popularity of each product line
    - Identify potential overstocking in lines with low order quantities
    - Ensure adequate variety in high-performing product lines
3. Adjust marketing strategy:
    - Focus promotional efforts on product lines with high potential (many products but lower order quantities)
    - Capitalize on the popularity of high-performing lines in marketing campaigns
4. Advise Product development:
    - Invest in expanding product ranges for lines with high order quantities
    - Consider phasing out or revamping product lines with low numbers of products and low order quantities

Hint: Think about how you can and might have to utilize SQL DISTINCT statement

Execute a SQl query that returns the information required to address this ask.

```python
# Replace None with your code
df_product_lines = pd.read_sql("""
SELECT pl.productLine,
       COUNT(DISTINCT p.productCode) AS num_products,
       SUM(od.quantityOrdered) AS total_quantity
FROM productlines pl
JOIN products p USING(productLine)
JOIN orderdetails od USING(productCode)
GROUP BY pl.productLine
ORDER BY total_quantity DESC
""", conn)
df_product_lines
```

Once you have the information returned in a dataframe, select an appropriate visualization to represent the relationship between total quantity ordered and the number of products in order to perform a preliminary investigation into the question of if more products lead to more orders. You are welcome to utilize matplotlib, seaborn, or pandas plotting to produce your visual. Ensure that it has a verbose title and axis labels!

```python
# Replace None with your visual code
# Produce a visual to represent the the relation between number of products and the total amount ordered
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df_product_lines, x='num_products', y='total_quantity', s=150)
for _, r in df_product_lines.iterrows():
    plt.annotate(r['productLine'], (r['num_products'], r['total_quantity']),
                 xytext=(6, 6), textcoords='offset points')
plt.title('Number of Products vs. Total Quantity Ordered by Product Line')
plt.xlabel('Number of Products in Line')
plt.ylabel('Total Quantity Ordered')
plt.tight_layout()
plt.show()
```

### Step 8: Remote Offices
Upper management is considering a shift to hybrid and remote work for certain locations and roles. They have tasked you with providing them data about employees who work in any office that has fewer than 5 total employees so they can better understand how to support those employees remotely when offices are shut down. 

Be sure to include information about the employees job and supervisor so management can adjust everyone to remote work properly.

Hint: Utilize a subquery to find the relevant offices

Execute a SQl query that returns the information required to address this ask.

```python
# Replace None with your code
df_small_offices = pd.read_sql("""
SELECT e.employeeNumber, e.firstName, e.lastName, e.jobTitle,
       e.officeCode, o.city,
       e.reportsTo,
       m.firstName AS supervisorFirstName,
       m.lastName AS supervisorLastName
FROM employees e
JOIN offices o USING(officeCode)
LEFT JOIN employees m ON e.reportsTo = m.employeeNumber
WHERE e.officeCode IN (
    SELECT officeCode
    FROM employees
    GROUP BY officeCode
    HAVING COUNT(*) < 5
)
ORDER BY e.officeCode, e.lastName
""", conn)
df_small_offices
```

## Reflection Question:

Describe how you decided on the subquery that you used in the query above? This answer can be technically in nature, describing your thought process in how the main query is utilizing the subquery to return the correct data.
**Answer:**

The employees table doesn't store how many people work in each office, so I used a subquery to count them. It groups employees by office and keeps only offices with fewer than 5 employees (using HAVING, since the condition is on a count), giving a list of office codes.

The main query then returns employees whose office code is IN that list, so only people from small offices appear. I also joined the offices table for the city, and joined the employees table to itself on reportsTo to show each person's supervisor.

#### movies per year

```python
# Begin your code here
import matplotlib.pyplot as plt

# Movies per year, keeping only 2010-2019
df_year_counts = pd.read_sql("""
SELECT start_year, COUNT(*) AS num_movies
FROM movie_basics
WHERE start_year BETWEEN 2010 AND 2019
GROUP BY start_year
ORDER BY start_year
""", conn4)
df_year_counts
```

#### checking for misssing values

```python
# How many nulls are in the key columns?
df_nulls = pd.read_sql("""
SELECT COUNT(*) AS total_movies,
       SUM(runtime_minutes IS NULL) AS null_runtime,
       SUM(genres IS NULL) AS null_genres
FROM movie_basics
WHERE start_year BETWEEN 2010 AND 2019
""", conn4)
df_nulls
```

#### Ratings and genrees with Nulls excluded

```python
# Ratings and genres for rated movies, with null genres excluded
df_ratings = pd.read_sql("""
SELECT mb.movie_id, mb.genres, mr.averagerating, mr.numvotes
FROM movie_basics mb
JOIN movie_ratings mr USING(movie_id)
WHERE mb.start_year BETWEEN 2010 AND 2019
  AND mb.genres IS NOT NULL
""", conn4)
df_ratings.head()
```

#### summary of the genre

```python
# Genres are stored as one comma-separated string, so split into one row per genre
df_genres = df_ratings.assign(genre=df_ratings['genres'].str.split(',')).explode('genre')

# Average rating and votes per genre, keeping genres with at least 500 movies
df_genre_summary = (df_genres.groupby('genre')
    .agg(num_movies=('movie_id', 'count'),
         avg_rating=('averagerating', 'mean'),
         avg_votes=('numvotes', 'mean'))
    .query('num_movies >= 500')
    .sort_values('avg_rating', ascending=False)
    .reset_index())
df_genre_summary
```

### The Visual

```python
# Visual: average rating vs. audience engagement by genre
fig, ax = plt.subplots(figsize=(11, 7))
ax.scatter(df_genre_summary['avg_votes'], df_genre_summary['avg_rating'], s=80)
for _, r in df_genre_summary.iterrows():
    ax.annotate(r['genre'], (r['avg_votes'], r['avg_rating']),
                xytext=(5, 5), textcoords='offset points')
ax.set_xscale('log')
ax.set_title('Average Rating vs. Average Audience Votes by Genre (2010-2019)')
ax.set_xlabel('Average Number of Votes per Movie (log scale)')
ax.set_ylabel('Average IMDB Rating')
plt.tight_layout()
plt.show()
```

## Data quality checks

#### Movies dated after 2019

```python
# Cleaning check 1: movies dated after 2019
df_after_2019 = pd.read_sql("""
SELECT COUNT(*) AS movies_after_2019, MAX(start_year) AS latest_year
FROM movie_basics
WHERE start_year > 2019
""", conn4)
df_after_2019
```

#### outliers in the Runtime

```python
# Cleaning check 2: runtime outliers (2010-2019)
df_runtime = pd.read_sql("""
SELECT MAX(runtime_minutes) AS max_runtime,
       SUM(runtime_minutes < 10) AS runtime_under_10
FROM movie_basics
WHERE start_year BETWEEN 2010 AND 2019
""", conn4)
df_runtime
```

#### ratings with few votes

```python
# Cleaning check 3: ratings based on fewer than 10 votes
df_low_votes = pd.read_sql("""
SELECT COUNT(*) AS ratings_under_10_votes
FROM movie_ratings
WHERE numvotes < 10
""", conn4)
df_low_votes
```

#### movie titles that are duplicates

```python
# Cleaning check 4: duplicate title and year pairs
df_duplicates = pd.read_sql("""
SELECT COUNT(*) AS duplicate_title_year
FROM (
    SELECT 1
    FROM movie_basics
    GROUP BY primary_title, start_year
    HAVING COUNT(*) > 1
)
""", conn4)
df_duplicates
```
